```python
from sklearn.datasets import make_regression
import numpy as np
import matplotlib.pyplot as plt 
```

```python
X,y = make_regression(n_samples=100, n_features=1 ,n_targets=1,noise=20)
plt.scatter(X,y)
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(X,y)

print("b = ",lr.intercept_)
print("m = ",lr.coef_)
```

```python
class GradientDescent:
    def __init__(self,lr,epochs):
        self.m = 100
        self.b = -120
        self.lr = lr
        self.epochs = epochs
        self.all_b = [100]
        self.all_m = [-120]
    def fit(self,X,y):
        for i in range (self.epochs):
            loss_Slope_b= -2 * np.sum(y - self.m *X.ravel() - self.b)
            loss_Slope_m = -2 * np.sum((y - self.m *X.ravel() - self.b)* X.ravel()) 
            self.b = self.b - self.lr * loss_Slope_b
            self.m= self.m - self.lr * loss_Slope_m  
            self.all_b.append(self.b)
            self.all_m.append(self.m)
        print("b= ",self.b)
        print("m= ",self.m)
    def predict(self,X):
        return self.m*X+self.b
    def getBM(self):
        return self.all_b ,self.all_m
    
```

```python
gd = GradientDescent(0.001,50)
gd.fit(X,y)
```

```python
print(gd.predict(X))
```

```python
all_b,all_m = gd.getBM()
plt.scatter(X,y)
for i in range(len(all_b)):
    plt.plot()
```
