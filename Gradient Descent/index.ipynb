```python
from sklearn.datasets import make_regression
import numpy as np
import matplotlib.pyplot as plt
```

```python
X,y = make_regression(n_samples=4,n_features=1,n_targets=1,noise=80,random_state=13)
```

```python
plt.scatter(X,y)
```

```python
#applying OLS
from sklearn.linear_model import LinearRegression
reg = LinearRegression()

reg.fit(X,y)

reg.coef_
```

```python
reg.intercept_
```

```python
plt.scatter(X,y)
plt.plot(X,reg.predict(X),color='red')
```

```python
#applying gradient descent assuming that m = 78.35 , 
#starting value of b = 0
y_pred = (78.35*X + 0).reshape(4)
```

```python
plt.scatter(X,y)
plt.plot(X,reg.predict(X),color='red')
plt.plot(X,y_pred ,color='blue')
```

```python
m = 78.35
b = 0

loss_Slope = -2 * np.sum(y - m *X.ravel() - b)
loss_Slope
```

```python
learning_rate = 0.1


step_size = learning_rate * loss_Slope
step_size
```

```python
b = b - step_size
b
```

```python
y_pred = (78.35*X + b).reshape(4)
```

```python
plt.scatter(X,y)
plt.plot(X,reg.predict(X),color='red')
plt.plot(X,y_pred ,color='blue')
```

```python
#lets iterate above steps
b = -100
m = 78.35
epochs = 10
learning_rate = 0.1 
for i in range(epochs):
    loss_Slope = -2 * np.sum(y - m *X.ravel() - b)
    b = b - (learning_rate*loss_Slope)
    y_pred = m*X+b
    plt.scatter(X,y)

    plt.plot(X,y_pred)
    
```
